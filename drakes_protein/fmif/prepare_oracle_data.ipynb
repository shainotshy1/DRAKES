```python
import pickle
from pathlib import Path
```

```python
data_path = Path("/u/sdickman/DRAKES/drakes_protein/fmif/align_oracle_data/")

all_data = []
for data_file in data_path.glob("*.pkl"):
    with open(data_file, "rb") as f:
        data = pickle.load(f)
        if isinstance(data, list):
            all_data.extend(data)
        else:
            all_data.append(data)

target_path = Path.joinpath(data_path, "full_traj_train_pretrained.pkl")
with open(target_path, "wb") as f:
    pickle.dump(all_data, f)

print(len(all_data))
```
